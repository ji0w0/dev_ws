```python
from ultralytics import YOLO

model = YOLO("/home/jiu/dev_ws/yolov11/runs/detect/~/yolo/myproject/train/weights/best.pt")
```

```python
results = model.predict(
    source="./download (8).jpeg",  # 이미지/폴더/영상 경로 (웹캠은 0)
    conf=0.25,       # confidence 임계값
    iou=0.7,         # NMS IoU 임계값
    imgsz=640,       # 추론 이미지 크기
    save=True,       # 결과 이미지를 runs/detect/predict*/ 에 저장
    save_txt=False,  # 라벨(txt) 저장 여부
    device=0,        # GPU 사용 (CPU는 "cpu")
)

for r in results:
    boxes = r.boxes
    for box in boxes:
        cls_id = int(box.cls[0])                 # 클래스 ID
        cls_name = model.names[cls_id]           # 클래스 이름
        conf = float(box.conf[0])                # 신뢰도
        x1, y1, x2, y2 = box.xyxy[0].tolist()    # 바운딩 박스 좌표
        print(f"{cls_name} ({conf:.2f}): [{x1:.0f}, {y1:.0f}, {x2:.0f}, {y2:.0f}]")

    r.show()
    r.save(filename="result.jpg")
```
