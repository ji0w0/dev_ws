```python
samples = [1, 7, 9, 16, 36, 39, 45, 45, 46, 48, 51, 100, 101]
tmp_y = [1]*len(samples)
tmp_y
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 4))
plt.scatter(samples, tmp_y)
plt.grid()
plt.show()
```

```python
import numpy as np

iqr = np.percentile(samples, 75) - np.percentile(samples, 25) # IQR = Q3 - Q1
iqr * 1.5
```

```python
q1 = np.percentile(samples, 25) # 1사분위수
q2 = np.median(samples) # 2사분위수
q3 = np.percentile(samples, 75) # 3사분위수
upper_fence = q3 + iqr*1.5
lower_fence = q1 - iqr*1.5
```

```python
plt.figure(figsize=(12,4))
plt.scatter(samples, tmp_y)
plt.axvline(x=q1, color='black') # 1사분위수
plt.axvline(x=q2, color='red') # 2사분위수
plt.axvline(x=q3, color='black') # 3사분위수
plt.axvline(x=upper_fence, color='black', ls='dashed') # 상한 경계선
plt.axvline(x=lower_fence, color='black', ls='dashed') # 하한 경계선
plt.grid()
plt.show()
```

```python
import seaborn as sns

plt.Figure(figsize=(12, 4))
sns.boxplot(x=samples)
plt.grid()
plt.show()
```
