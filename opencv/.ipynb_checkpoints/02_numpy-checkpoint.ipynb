```python
import numpy as np # 다차원 배열 처리
import matplotlib.pyplot as plt # 시각화
from PIL import Image
```

```python
pic = Image.open('./data/test.png')
pic
```

```python
type(pic)
```

```python
pic_arr = np.asarray(pic)
type(pic_arr)
```

```python
pic_arr.shape # 세로, 가로, 채널(rgb)
```

```python
plt.imshow(pic_arr)
```

```python
pic_red = pic_arr.copy()
plt.imshow(pic_red[:, :, 0], cmap="gray") # 0번 채널만
```

```python
pic_green = pic_arr.copy()
plt.imshow(pic_green[:, :, 1], cmap="gray") # 1번 채널만
```

```python
pic_red[:, :, 1] = 0 # Green 채널의 모든 값을 0으로 변경
plt.imshow(pic_red);
```

```python
pic_red[:, :, 2] = 0 # Blue 채널의 모든 값을 0으로 변경
plt.imshow(pic_red);
```
