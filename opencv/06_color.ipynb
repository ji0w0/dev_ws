```python
img2 = cv2.imread('./data/watermark_no_copy.png')
img2 = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
plt.imshow(img2)
```

```python
img1.shape, img2.shape
```

```python
img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))
img1.shape, img2.shape
```

```python
blended = cv2.addWeighted(src1=img1, alpha=0.5, src2=img2, beta=0.5, gamma=0)
plt.imshow(blended)
```

```python
img2 = cv2.resize(img2, (600, 600))
plt.imshow(img2)
```

```python
large_img = img1
small_img = img2

x_offset = 0
y_offset = 0
```

```python
x_end = x_offset + small_img.shape[1]
y_end = y_offset + small_img.shape[0]

large_img[y_offset:y_end, x_offset:x_end] = small_img

plt.imshow(large_img)
```

```python
img1 = cv2.imread('./data/sample.jpg')
img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2 = cv2.imread('./data/watermark_no_copy.png')
img2 = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
img2 = cv2.resize(img2, (600, 600))

img1.shape, img2.shape
```

```python
x_offset = img1.shape[1] - img2.shape[1]
y_offset = img1.shape[0] - img2.shape[0]
x_offset, y_offset
```

```python
roi = img1[y_offset : img1.shape[0], x_offset : img1.shape[1]]
plt.imshow(roi) # region of interest
```

```python
img2_gray = cv2.cvtColor(img2, cv2.COLOR_RGB2GRAY)
plt.imshow(img2_gray, cmap='gray')
```

```python
mask_inv = cv2.bitwise_not(img2_gray)
plt.imshow(mask_inv, cmap='gray')
```

```python
mask_inv.shape
```

```python
fg = cv2.bitwise_or(img2, img2, mask=mask_inv) # mask의 모양대로 img2의 특정 영역만 남김
plt.imshow(fg)
```

```python
final_roi = cv2.bitwise_or(roi, fg)
plt.imshow(final_roi)
```

```python
img1[y_offset : y_offset + img2.shape[0], x_offset : x_offset + img2.shape[1]] = (
    final_roi
)
plt.imshow(img1)
```
